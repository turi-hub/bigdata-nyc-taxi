# 02 — Modelo Batch con SparkML

### 02.1 LIMPIEZA DE DATOS

```python
from pyspark.sql import SparkSession

try:
    spark.stop()
except NameError:
    pass

spark = (
    SparkSession.builder
    .appName("nyc-taxi-ml")
    .master("local[*]")
    .config("spark.executor.memory", "4g")
    .config("spark.driver.memory", "4g")
    .config("spark.sql.execution.arrow.pyspark.enabled", "false")
    .config("spark.sql.execution.arrow.pyspark.fallback.enabled", "false")
    .getOrCreate()
)

print("Arrow enabled:", spark.conf.get("spark.sql.execution.arrow.pyspark.enabled"))
```

```python
df = spark.read.parquet("../data/*.parquet")
df.printSchema()
```

Revisamos registros que no tengan sentido, como viajes con distancia demasiado pequeña, fechas de llegada anteriores a recogida, tarifas fuera de rango,...

```python
from pyspark.sql.functions import col, count, when

# Total de registros
print(f"Total registros: {df.count()}")

# Cuántos incumplen cada condición por separado
df.select(
    count(when(col("Trip_Distance") <= 0.1, 1)).alias("distancia_invalida"),
    count(when(col("Fare_Amt") <= 0, 1)).alias("tarifa_invalida"),
    count(when(col("Trip_Dropoff_Datetime") <= col("Trip_Pickup_Datetime"), 1)).alias("tiempo_invalido"),
).show()
```

Observamos 185877 registros con distancia menos a 0.1 millas y 259749 registros con fecha de llegada anterior a fecha de recogida
- Teniendo en cuenta que representan menos del 2% del conjunto de datos, sera bastante probable que eliminemos

```python
df_clean = df.filter(
    (col("Trip_distance") > 0.1) &
    (col("Trip_Dropoff_Datetime") > col("Trip_Pickup_Datetime"))
)

print(f"Registros originales:  {df.count()}")
print(f"Registros tras filtro: {df_clean.count()}")
```

Ahora miramos posibles outliers
- Hemos de tener en cuenta que si predecimos demanda por zona, no todas las variables nos interesan por lo que el estudio de todos los outliers tampoco es necesario. 
- Las variables que nos interesan para predecir la demanda; Trip_Pickup_DateTime (y las que derivemos de ella), Zona (derivada de Start_Lon y Start_Lat) y Trip_Count (que debemos generar para ver el numero de viajes por zona en ventana temporal de X minutos)

```python
from pyspark.sql.functions import col, min, max
df_clean.select(
    min("Start_Lon").alias("lon_min"),
    max("Start_Lon").alias("lon_max"),
    min("Start_Lat").alias("lat_min"),
    max("Start_Lat").alias("lat_max")
).show()
```

Observamos que hay coordenadas que estan muy lejos de ser NY vamos a establecer un cuadrado para delimitar lo que consideramos NY

```python
n_fuera_nyc = df_clean.filter(
    (col("Start_Lon") < -74.3) | (col("Start_Lon") > -73.7) |
    (col("Start_Lat") < 40.5)  | (col("Start_Lat") > 40.9)
).count()

print(f"Registros fuera de NYC: {n_fuera_nyc} ({n_fuera_nyc / df_clean.count() * 100:.2f}%)")
```

```python
df_fuera = df_clean.filter(
    (col("Start_Lon") < -74.3) | (col("Start_Lon") > -73.7) |
    (col("Start_Lat") < 40.5)  | (col("Start_Lat") > 40.9)
)

df_fuera.select("Start_Lon", "Start_Lat").describe().show()
# Ver si hay (0,0) u otros valores absurdos
df_fuera.filter((col("Start_Lon") == 0) | (col("Start_Lat") == 0)).count()
```

Vemos que de hecho de 233722 registros, 205105 son valores (0,0) que son coordenadas corruptas y no limitrofes con las que hemos seleccionado, siendo un porcentaje muy alto, y los registros totales solo un 2% vamos a decidir eliminarlos.

```python
print(f"Registros antes: {df_clean.count()}")
df_clean = df_clean.filter(
    (col("Start_Lon") >= -74.3) & (col("Start_Lon") <= -73.7) &
    (col("Start_Lat") >= 40.5)  & (col("Start_Lat") <= 40.9)
)
print(f"Registros despues: {df_clean.count()}")
```

Ahora analizamos los tiempos de recogida y salida de los clientes para detectar outliers

```python
from pyspark.sql.functions import min, max

df_clean.select(
    min("Trip_Pickup_DateTime"),
    max("Trip_Pickup_DateTime")
).show()
```

```python
from pyspark.sql.functions import min, max

df_clean.select(
    min("Trip_Dropoff_DateTime"),
    max("Trip_Dropoff_DateTime")
).show()
```

- Vemos que el maximo de Dropoff esta en marzo, revisamos

```python
df_clean.filter(col("Trip_Dropoff_DateTime") > "2009-01-31").count()
```

- Puede ser que al haber horas de recogida el 31 de enero a las 23:59 las de llegada sean el 1 de febrero. Pero si estan mas lejos de ahi, supondremos outlier al entender que un viaje en taxi no durara mas de un dia

```python
from pyspark.sql.functions import unix_timestamp

df_clean.filter(
    (unix_timestamp("Trip_Dropoff_DateTime") - unix_timestamp("Trip_Pickup_DateTime")) / 3600 > 24).count()
```

```python
df_clean.filter(
    (unix_timestamp("Trip_Dropoff_DateTime") - unix_timestamp("Trip_Pickup_DateTime")) / 3600 > 24).show(50, truncate = False)
```

No solo hemos visto que hay viajes pasando febrero si no que hay 48 viajes de mas de 24 horas de duracion que vamos a tener que tratar como outliers si entendemos que un viaje no puede durar mas de 24 horas. Por curiosidad miramos cuantos viajes hay con mas de 2,4,6,8,10 y 12 horas de duracion

```python
for horas in [2,4,6,8,10,12]: 
    n = df_clean.filter(
        (unix_timestamp("Trip_Dropoff_DateTime") - unix_timestamp("Trip_Pickup_DateTime")) / 3600 > horas).count()
    print(f"Mas de {horas}h: {n:,} ({n / df_clean.count() * 100:.2f}%)")
```

- Debemos pensar que umbral elegimos como valido para un viaje de taxi...
